```python
import pandas as pd

FILE = r"C:\Users\Audrey\Python Projects\Gateway\Data\Gateway Graphs.xlsx"

plastic = pd.read_excel(FILE, sheet_name="Plastic Graph")
disposables = pd.read_excel(FILE, sheet_name="Disposables Graph")
diversion = pd.read_excel(FILE, sheet_name="C&D Diversion Data")
```

```python
plastic
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    plastic["Year"],
    plastic["Plastic Prevented (kg)"],
    marker="o",
    markersize=9,
    linewidth=3
)
plt.title(
    "Projected Plastic Waste Prevented Through\nRefillable Office Supplies",
    fontsize=15,
    pad=22
)
plt.xlabel("Year", fontsize=13)
plt.ylabel("Plastic Prevented (kg)", fontsize=13)
plt.xticks(plastic["Year"])

plt.ylim(35,120)

plt.grid(axis="y", alpha=0.25)

ax = plt.gca()

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()

plt.savefig(
    r"C:\Users\Audrey\Python Projects\Gateway\Figures\Plastic_Prevented.png",
    dpi=300,
    bbox_inches="tight"
)

plt.savefig(
    r"C:\Users\Audrey\Python Projects\Gateway\Figures\Plastic_Prevented.svg",
    bbox_inches="tight"
)

plt.savefig(
    r"C:\Users\Audrey\Python Projects\Gateway\Figures\Plastic_Prevented.pdf",
    bbox_inches="tight"
)

plt.show()
```

```python
from matplotlib.ticker import StrMethodFormatter

plt.figure(figsize=(8,5))

plt.plot(
    disposables["Year"],
    disposables["Single Use Items Eliminated"],
    marker="o",
    markersize=9,
    linewidth=3
)

plt.title(
    "Projected Single-Use Items Eliminated\nThrough Social Kitchen Reusables",
    fontsize=15,
    pad=22
)

plt.xlabel("Year", fontsize=13)
plt.ylabel("Single-Use Items Eliminated", fontsize=13)
plt.xticks(disposables["Year"])

plt.ylim(250000, 1300000)

plt.grid(axis="y", alpha=0.25)

ax = plt.gca()
ax.yaxis.set_major_formatter(StrMethodFormatter("{x:,.0f}"))

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()


plt.savefig(r"C:\Users\Audrey\Python Projects\Gateway\Figures\Single_Use_Items_Eliminated.svg", bbox_inches="tight")
plt.savefig(r"C:\Users\Audrey\Python Projects\Gateway\Figures\Single_Use_Items_Eliminated.png", dpi=300, bbox_inches="tight")
plt.savefig(r"C:\Users\Audrey\Python Projects\Gateway\Figures\Single_Use_Items_Eliminated.pdf", bbox_inches="tight")

plt.show()
```

```python
# Keep only materials with at least 25 tons
diversion = diversion[diversion["Tons Diverted"] >= 25]

# Sort for horizontal bar chart
diversion = diversion.sort_values("Tons Diverted", ascending=True)

plt.figure(figsize=(9, 5.5))

bars = plt.barh(
    diversion["Material"],
    diversion["Tons Diverted"]
)

plt.title(
    "Construction Materials Diverted From Landfill\nDuring Gateway Construction",
    fontsize=16,
    pad=18
)

plt.xlabel("Material Recovered (Tons)", fontsize=13)
plt.ylabel("")
plt.grid(axis="x", alpha=0.3)

# For values displayed on bars
for i, value in enumerate(diversion["Tons Diverted"]):
    plt.text(
        value + 40,
        i,
        f"{value:,.0f}",
        va="center",
        fontsize=10
    )

# Remove top and right border
ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.savefig(r"C:\Users\Audrey\Python Projects\Gateway\Figures\Construction_Materials_Diverted.svg", bbox_inches="tight")
plt.savefig(r"C:\Users\Audrey\Python Projects\Gateway\Figures\Construction_Materials_Diverted.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```
